# Claude API Starter Notebook

Authentication uses a LiteLLM proxy, so we need `base_url` + `ANTHROPIC_AUTH_TOKEN` instead of the standard `ANTHROPIC_API_KEY`.

```python
import os
import anthropic
from dotenv import load_dotenv

load_dotenv()  # loads variables from .env in this directory

client = anthropic.Anthropic(
    base_url=os.environ.get("ANTHROPIC_BASE_URL"),
    api_key=os.environ.get("ANTHROPIC_AUTH_TOKEN"),
)

MODEL = os.environ.get("ANTHROPIC_MODEL", "anthropic/claude-sonnet-4-6")

print(f"Using model: {MODEL}")
print(f"Base URL:    {os.environ.get('ANTHROPIC_BASE_URL')}")
```

## Basic message

```python
message = client.messages.create(
    model=MODEL,
    max_tokens=1024,
    messages=[
        {"role": "user", 
         "content": "Hello, Claude!"}
    ]
)

print(message.content[0].text)
```

## With a system prompt

```python
message = client.messages.create(
    model=MODEL,
    max_tokens=1024,
    system="You are a helpful assistant that responds concisely.",
    messages=[
        {"role": "user", "content": "What is the capital of France?"}
    ]
)

print(message.content[0].text)
```

## Multi-turn conversation

```python
messages = [
    {"role": "user", "content": "My name is Diego."},
]

response = client.messages.create(model=MODEL, max_tokens=256, messages=messages)
reply = response.content[0].text
print(f"Claude: {reply}")

# Append assistant reply and continue the conversation
messages.append({"role": "assistant", "content": reply})
messages.append({"role": "user", "content": "What is my name?"})

response2 = client.messages.create(model=MODEL, max_tokens=256, messages=messages)
print(f"Claude: {response2.content[0].text}")
```
